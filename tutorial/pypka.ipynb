```python
from pypka import Titration
from propka import run
import pandas as pd
import numpy as np
```

```python
params = {
 'structure'     : '/home/ivan/Sarcomere/Lysozyme/TrIPP_test/TrIPP/tutorial/output/1AKI_md1_300ns/.temp_572842.pdb', # Input structure file name
 'ncpus'         : -1,         # Number of processes (-1 for maximum allowed)
 'epsin'         : 15,         # Dielectric constant of the protein
 'ionicstr'      : 0.1,        # Ionic strength of the medium (M)
 'pbc_dimensions': 0           # PB periodic boundary conditions (0 for solvated proteins and 2 for lipidic systems)
}
tit = Titration(params)

# pH = 1.0
# for site in tit:
#     state = site.getProtState(pH)[0]
#     print(site.res_name, site.res_number, site.pK, state)
```

```python
processed_line = []
for line in str(tit).split('\n'):
    tmp = []
    if 'Not In Range' in line or 'Predicted Isoelectric Point' in line:
        continue
    for element in line.split(' '):
        if element != '':
            tmp.append(element)
    if tmp:
        processed_line.append(tmp)
```

```python
df = pd.DataFrame(processed_line[1:], columns=processed_line[0])
```
